```python

import pandas as pd
from io import StringIO

# Student dataset in CSV format
csv_data = """student_id,name,gender,study_hours,marks
101,Amit,Male,2,65
102,Priya,Female,4,82
103,Rahul,Male,3,75
104,Sneha,Female,5,90
105,Rohan,Male,,68
106,Neha,Female,6,95
107,Karan,Male,2,
108,Pooja,Female,4,85
109,Arjun,Male,3,72
102,Priya,Female,4,82
"""

# Load dataset into Pandas
df = pd.read_csv(StringIO(csv_data))

print("Original Data:")
print(df)
print(df.info())

```

```python
df["gender"]=["Male","Female","Male","Female","Male","Female","Male","Female","Male","Female"]
df
```

```python
print(df.info())
```

```python
df["study_hours"]=pd.to_numeric(df["study_hours"],errors="coerce")
```

```python
#Remove duplicates based on all columns(or key columns like student_id, name)
df=df.drop_duplicates(subset=["student_id","name"], keep="first")
df
```

```python
#handle missing values
#for numeric columns filled with median.
df["study_hours"]=df["study_hours"].fillna(df["study_hours"].median())
df
```

```python
df["marks"]=df["marks"].fillna(df["marks"].median())
df
```

```python
#Basic statistics
print("Basic Statistics: ")
print(df[["study_hours","marks"]].describe())
```

```python
#Mean, Median
print("Mean: ")
print(df[["study_hours","marks"]].mean())
print(df[["study_hours","marks"]].median())
```

```python
#Value counts for categorial column
print(df["gender"].value_counts())
df
```
